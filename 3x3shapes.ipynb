```python
import numpy as np
import plotly.graph_objects as go 

def jk(vecs):
    vecs_c = vecs.copy()
    d = np.array(vecs).sum(axis = 0)
    vecs_c.append(d)
    z = np.array([0, 0, 0])
    vecs_c.append(z)
    for i in range(len(vecs)):
        for j in range(i+1, len(vecs)):
            h = vecs[i] + vecs[j]
            vecs_c.append(h)
    return vecs_c


def draw3x3(A):
    v1 = A[:, 0]
    v2 = A[:, 1]
    v3 = A[:, 2]

    fig = go.Figure()

    def add_vec(fig, v):
        fig.add_trace(go.Scatter3d(
            x = [0, v[0]],
            y = [0, v[1]],
            z = [0, v[2]],
            mode = 'lines+markers+text', line = dict(color = 'red')

        ))


    add_vec(fig, v1)
    add_vec(fig, v2)
    add_vec(fig, v3)

    lst = jk([v1, v2, v3])

    for i in range(3, len(lst)):
        fig.add_trace(go.Scatter3d(
        x = [0, (lst[i])[0]],
        y = [0, (lst[i])[1]],
        z = [0, (lst[i])[2]],
        mode = 'markers', marker = dict(color = 'red'), text = [None, 'name']))


    for i, j in enumerate(lst, 0):
        newlst = lst.copy()
        _ = newlst.pop(i)
        for k in newlst:

            fig.add_trace(go.Scatter3d(
                x = [j[0], k[0]],
                y = [j[1], k[1]],
                z = [j[2], k[2]],
                mode = 'lines', line = dict(color = 'red'), text = [None, 'name']))



    def axis(fig, l):
        fig.add_trace(go.Scatter3d(
            x = [-l, l],
            y = [0, 0],
            z = [0, 0],
            mode = 'lines+text', line = dict(width = 3, color = 'black'), text = 'x'
        ))
        fig.add_trace(go.Scatter3d(
            x = [0, 0],
            y = [-l, l],
            z = [0, 0],
            mode = 'lines+text', line = dict(width = 3, color = 'black'), text = 'y'
        ))
        fig.add_trace(go.Scatter3d(
            x = [0, 0],
            y = [0, 0],
            z = [-l, l],
            mode = 'lines+text', line = dict(width = 3, color = 'black'), text = 'z'
        ))
    axis(fig, 10)

    fig.show()


de = np.eye(3, 3)
np.random.seed(29)
f = np.random.randint(1, 8, (3, 3))
g = np.eye(3, 3) * 5

x = np.array([[2, 3, 0],
              [1, 4, 5],
              [0, 2, 0]])

y = np.array([[2, 3, 0],
              [-4, -1, 0],
              [0, 0, 0]])

e = np.eye(3, 3)
aa = np.array([[1, 0, 0],
               [0, 1, 0],
               [0, 0, 1]]) * 5
draw3x3(aa)


```
